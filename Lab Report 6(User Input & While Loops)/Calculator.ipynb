```python
def add(x,y):
    return x+y
def sub(x,y):
    return x-y
def mul(x,y):
    return x*y
def div(x,y):
    if y!=0:
        return x/y
    else:
        print("Can't be divided")
def make_caclculator(asmd,a,b):
    result=asmd(a,b)
    return result
print("Add: ", make_caclculator(add,45,8))
print("Sub : ", make_caclculator(sub,58,42))
print("Multiplication: " , make_caclculator(mul,32,8))
print("Division: ", make_caclculator(div,84,4))
```
