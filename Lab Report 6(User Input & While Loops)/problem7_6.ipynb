```python
prompt="Enter a pizza topping :\n"
prompt +="Enter 'quit' to end the program. "
topping=""
while topping!='quit':
    topping=input(prompt)
    if topping!='quit':
     print(f"I'll add {topping.title()} to your pizza.")
```

```python
active=True
while active:
    topping=input(prompt)
    if topping=='quit':
        active=False
    else:
     print(f"I'll add {topping.title()} to your pizza.")

```
