```python
prompt="Enter your age : \n"
prompt +="\nEnter 'quit' to end the program. "
while True:
    age=input(prompt)
    if age=='quit':
        break
    age=int(age)
    if age<3:
        print("Ticket is Free!")
    elif age>3 and age<=12:
        print("The ticket is $10.")
    elif age>12:
        print("The ticket is $10.")
```
