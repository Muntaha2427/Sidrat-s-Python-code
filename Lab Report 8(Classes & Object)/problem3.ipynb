```python
class Person:
    def __init__(self,name,nationality,age):
        self.name=name
        self.nationality=nationality
        self.age=age
    def details(self):
        print("Person's Details:")
        print("Name : ", self.name)
        print("Nationality : ",self.nationality)
        print("Age : ", self.age)
detail=Person("Sidratul Muntaha","Bangladeshi",20)
detail.details()
```
